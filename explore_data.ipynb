```python
import pyspark.sql.functions as f

df_weather = spark.sql(
    """select city_name, country_code, date, day_flag, minutes_of_precipitation_total, humidity_relative_avg
from samples.accuweather.historical_daynight_metric
-- where minutes_of_precipitation_total > 0 and humidity_relative_avg is not null
""")

df_weather = df_weather.withColumn('humidity', f.when(f.col('humidity_relative_avg') > 0, '>0')
                                                    .otherwise('0')
                                   )\
                        .withColumn('precipitation', f.when(f.col('minutes_of_precipitation_total') > 0, '>0')
                                                    .otherwise('0')
                                   )   

# df_g = df_weather.groupBy('country_code','date','humidity','precipitation').agg(f.count('*').alias('count'))

# display(df_g)

display(df_weather)
```
